**Import data dan libraries**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from transformers import AutoTokenizer, AutoModel
```

```python
df = pd.read_excel('scrapingTwitter.xlsx', engine='openpyxl')
```

**Preprocessing**

```python
import re
kolom_dipakai = ['id', 'text']
kolom_ada = [col for col in kolom_dipakai if col in df.columns]
df_clean = df[kolom_ada].copy()
```

```python
df_clean.head()
```

```python
#CLEANING NOISE TEKS
def clean_noise(text):
    if not isinstance(text, str):
        return ""

    # a. Hapus URL (http, https, www)
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)

    # b. Hapus Mention Twitter (@username)
    text = re.sub(r'@\w+', '', text)

    # c. Hapus Karakter Newline/Enter (\n, \r)
    text = re.sub(r'[\r\n]+', ' ', text)

    # d. Hapus Whitespace/Spasi berlebih (termasuk spasi di awal/akhir)
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
# Terapkan fungsi pembersihan ke kolom 'text'
df_clean['text_clean'] = df_clean['text'].apply(clean_noise)
```

```python
# Hapus baris jika ada teks yang kosong setelah dibersihkan
df_clean = df_clean[df_clean['text_clean'] != ''].reset_index(drop=True)
```

```python
# Tampilkan hasil
print("Ukuran dataset setelah drop kolom:", df_clean.shape)
print(df_clean[['text', 'text_clean']].head())
```

**Preprocessing case folding**

```python
def case_folding(text):
    return text.lower()

# Terapkan Tahap 2 (mengambil input dari hasil cleaning noise)
df_clean['step2_casefolding'] = df_clean['text_clean'].apply(case_folding)
```

```python
print(df_clean[['text_clean', 'step2_casefolding']].head())
```

**preprocessing normalisasi slang**

```python
# Buat kamus kata gaul / singkatan
slang_dict = {
    "bgt": "banget",
    "yg": "yang",
    "gw": "saya",
    "gua": "saya",
    "lu": "kamu",
    "pake": "pakai",
    "krn": "karena",
    "gk": "tidak",
    "ga": "tidak",
    "gak": "tidak",
    "sampe": "sampai",
    "tp": "tapi"
}

def normalize_slang(text):
    words = text.split()
    # Ganti kata jika ada di kamus, jika tidak ada biarkan tetap
    normalized_words = [slang_dict.get(word, word) for word in words]
    return ' '.join(normalized_words)

# Terapkan Tahap 3 (mengambil input dari hasil step2_casefolding)
df_clean['step3_normalized'] = df_clean['step2_casefolding'].apply(normalize_slang)
```
